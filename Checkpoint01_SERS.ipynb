## Dataset 1 — Appliances Energy Prediction (UCI)



##### Fonte: https://archive.ics.uci.edu/dataset/374/appliances+energy+prediction

**Situação**

Uma empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

```python
#Realizar a importacao do arquivo
dados = pd.read_csv('exercicio1_sample.csv')

# 1. Apresente head(), shape, info() e describe()
print("--- Primeiras linhas ---")
display(dados.head())
print("\n--- Dimensões (Linhas, Colunas) ---")
print(dados.shape)
print("\n--- Estrutura e Tipos de Dados ---")
dados.info()
print("\n--- Estatísticas Descritivas ---")
display(dados.describe())
```

```python
# 2. Renomeie Appliances para Consumo_Eletrodomesticos e simplifique atributos ambientais
dados = dados.rename(columns={
    "Appliances": "Consumo_Eletrodomesticos",
    "T1": "Temp_Cozinha",
    "RH_1": "Umid_Cozinha",
    "T2": "Temp_Sala",
    "RH_2": "Umid_Sala",
    "T3": "Temp_Lavabo",
    "RH_3": "Umid_Lavabo",
})

print(dados.columns)
```

```python
# 3. Determine o maior consumo de eletrodomésticos registrado na amostra
consumo_maximo = dados['Consumo_Eletrodomesticos'].max()
print(f"Maior consumo registrado: {consumo_maximo} Wh")

# 4. Calcule um limiar correspondente a 70% do valor máximo
limiar_70 = 0.70 * consumo_maximo
print(f"Limiar de 70% do máximo: {limiar_70} Wh")
df_alto_consumo = dados[dados['Consumo_Eletrodomesticos'] > limiar_70]

# 5. Conte os registros selecionados e calcule qual percentual da amostra eles representam
qtd_alto_consumo = len(df_alto_consumo)
percentual_alto = (qtd_alto_consumo / len(dados)) * 100

print(f"Quantidade de registros acima de 70%: {qtd_alto_consumo}")
print(f"Percentual que representam da amostra: {percentual_alto:.2f}%")
```

```python
# 6. Calcule a temperatura média de T1 (que renomeamos para Temp_Cozinha)
temp_media = dados['Temp_Cozinha'].mean()
print(f"Temperatura média (T1): {temp_media:.2f}°C")

df_consumo_e_temp = dados[
    (dados['Consumo_Eletrodomesticos'] > limiar_70) &
    (dados['Temp_Cozinha'] > temp_media)
]

qtd_consumo_e_temp = len(df_consumo_e_temp)
print(f"Registros com alto consumo E temperatura acima da média: {qtd_consumo_e_temp}")
```

Interpretação e Comparação dos Resultados:

DataFrame 1 (Apenas Consumo > 70%): Encontramos [inserir qtd_alto_consumo] registros que representam [inserir percentual_alto]% da amostra.

DataFrame 2 (Consumo > 70% + Temp T1 > Média): O número de registros caiu para [inserir qtd_consumo_e_temp].

Efeito da inclusão da temperatura: A introdução do segundo critério reduziu o tamanho do DataFrame resultante. Isso indica que nem todos os momentos de pico de consumo de eletrodomésticos ocorrem em condições de temperatura ambiente acima da média.
```python
# Carregar a amostra criada na Etapa A
dados = pd.read_csv('exercicio2steel_sample.csv')

# Renomear as colunas solicitadas
dados = dados.rename(columns={
    'Usage_kWh': 'Consumo_kWh',
    'Lagging_Current_Power_Factor': 'Fator_Potencia_Atrasado',
    'Leading_Current_Power_Factor': 'Fator_Potencia_Adiantado'
})

print(dados.columns)
```

```python
# Inspeção inicial: head, shape, info e describe
print("--- Primeiras linhas ---")
display(dados.head())

print("\n--- Dimensões (Linhas, Colunas) ---")
print(dados.shape)

print("\n--- Estrutura e Tipos de Dados ---")
dados.info()

print("\n--- Estatísticas Descritivas ---")
display(dados.describe())
```

```python
# Encontrar o maior consumo registrado
consumo_maximo = dados['Consumo_kWh'].max()

# Calcular 75% do consumo máximo
limiar_consumo = consumo_maximo * 0.75

print(f"Maior consumo registrado: {consumo_maximo:.2f} kWh")
print(f"Limiar de 75% do consumo máximo: {limiar_consumo:.2f} kWh")
```

```python
# Criar DataFrame com registros acima de 75% do consumo máximo
alto_consumo = dados[
    dados['Consumo_kWh'] > limiar_consumo
].copy()

quantidade_alto_consumo = len(alto_consumo)

percentual_alto_consumo = (
    quantidade_alto_consumo / len(dados)
) * 100

print("Quantidade de registros com consumo elevado:", quantidade_alto_consumo)
print(f"Percentual da amostra: {percentual_alto_consumo:.2f}%")

display(alto_consumo)
```

```python
# Normalizar os textos para funcionar tanto com
# "Maximum_Load" quanto com "Maximum Load"
load_type_normalizado = (
    alto_consumo['Load_Type']
    .astype(str)
    .str.replace('_', ' ', regex=False)
    .str.strip()
    .str.lower()
)

quantidade_maximum_load = (
    load_type_normalizado == 'maximum load'
).sum()

print(
    "Registros de alto consumo classificados como Maximum Load:",
    quantidade_maximum_load
)

if quantidade_alto_consumo > 0:
    percentual_maximum_load = (
        quantidade_maximum_load / quantidade_alto_consumo
    ) * 100

    print(
        f"Percentual dos registros de alto consumo que são Maximum Load: "
        f"{percentual_maximum_load:.2f}%"
    )
```

```python
# Vamos analisar o Fator de Potência Atrasado
print("--- Estatísticas do Fator de Potência Atrasado ---")
print(dados['Fator_Potencia_Atrasado'].describe())

print("\n--- Menores valores encontrados ---")
print(
    dados['Fator_Potencia_Atrasado']
    .sort_values()
    .head(20)
    .to_string(index=False)
)
```

```python
# Utilizar o primeiro quartil (25%) como limite
# para representar os valores mais baixos de fator de potência
limite_fator_potencia = dados[
    'Fator_Potencia_Atrasado'
].quantile(0.25)

print(
    f"Limite escolhido para fator de potência baixo: "
    f"{limite_fator_potencia:.2f}%"
)
```

```python
# Criar DataFrame com as duas condições simultaneamente:
# 1. Consumo acima de 75% do máximo
# 2. Fator de potência abaixo do limite escolhido

alto_consumo_fp_baixo = dados[
    (dados['Consumo_kWh'] > limiar_consumo) &
    (dados['Fator_Potencia_Atrasado'] < limite_fator_potencia)
].copy()

quantidade_consumo_fp_baixo = len(alto_consumo_fp_baixo)

percentual_consumo_fp_baixo = (
    quantidade_consumo_fp_baixo / len(dados)
) * 100

print(
    "Quantidade de registros com consumo elevado e fator de potência baixo:",
    quantidade_consumo_fp_baixo
)

print(
    f"Percentual da amostra: "
    f"{percentual_consumo_fp_baixo:.2f}%"
)

display(alto_consumo_fp_baixo)
```

```python
print("--- Interpretação e Comparação dos Resultados ---\n")

print(
    f"DataFrame 1 (Consumo > 75% do máximo): "
    f"Foram encontrados {quantidade_alto_consumo} registros, "
    f"representando {percentual_alto_consumo:.2f}% da amostra."
)

print(
    f"\nDesses registros, {quantidade_maximum_load} pertencem "
    f"à categoria Maximum Load."
)

print(
    f"\nFoi utilizado o Fator de Potência Atrasado e definido como "
    f"baixo um valor inferior a {limite_fator_potencia:.2f}%, "
    f"correspondente ao primeiro quartil da amostra."
)

print(
    f"\nDataFrame 2 (Consumo elevado + fator de potência baixo): "
    f"Foram encontrados {quantidade_consumo_fp_baixo} registros, "
    f"representando {percentual_consumo_fp_baixo:.2f}% da amostra."
)

print(
    "\nEsses registros merecem maior atenção da equipe de energia porque "
    "combinam um nível elevado de consumo com valores relativamente baixos "
    "de fator de potência. Essa combinação pode indicar momentos em que a "
    "energia elétrica está sendo utilizada de forma menos eficiente, com "
    "maior participação de potência reativa. Portanto, esses períodos são "
    "bons candidatos para investigação e possíveis ações de melhoria na "
    "eficiência energética da indústria."
)
```

## Dataset 3 — Power Consumption of Tetouan City (UCI)


#####Fonte: https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city

**Situação**

Uma equipe responsável pelo planejamento da distribuição de energia em Tétouan deseja identificar qual das três zonas apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda.


```python
# Carregar conjunto de dados
df = pd.read_csv('exercicio3_sample.csv')
```

```python
# Remova metadados/tipo linhas geradas pelo Orange Data Mining, se presente
df = df[pd.to_numeric(df['Temperature'], errors='coerce').notna()].copy()
```

```python
# Converter colunas numéricas em float
for col in df.columns:
    df[col] = pd.to_numeric(df[col])
```

```python
# 1. Renomeação das colunas
df.rename(
    columns={
        'PowerConsumption_Zone1': 'Consumo_Zona_1',
        'PowerConsumption_Zone2': 'Consumo_Zona_2',
        'PowerConsumption_Zone3': 'Consumo_Zona_3',
    },
    inplace=True,
)

print('=== 1. Renomeação de Colunas ===')
print('Colunas renomeadas com sucesso para:')
print(' - Consumo_Zona_1\n - Consumo_Zona_2\n - Consumo_Zona_3\n')
```

```python
# 2. Consumo Máximo por Zona
max_z1 = df['Consumo_Zona_1'].max()
max_z2 = df['Consumo_Zona_2'].max()
max_z3 = df['Consumo_Zona_3'].max()

print('=== 2. Consumo Máximo por Zona ===')
print(f'Consumo_Zona_1: {max_z1:,.2f} kW')
print(f'Consumo_Zona_2: {max_z2:,.2f} kW')
print(f'Consumo_Zona_3: {max_z3:,.2f} kW\n')
```

```python
# 3. Maior Pico de Consumo
print('=== 3. Maior Pico de Consumo ===')
print(
    f'Maior Pico Global: Consumo_Zona_1 com o pico de {max_z1:,.2f} kW.\n'
)
```

```python
# 4. Limiar de 70%
limiar_70 = 0.70 * max_z1

print('=== 4. Cálculo do Limiar de 70% ===')
print(f'Calculo: 0,70 * {max_z1:,.2f} kW')
print(f'Limiar de 70%: {limiar_70:,.2f} kW\n')
```

```python
# 5. Contagem e Percentual (DataFrame 1)
df1 = df[df['Consumo_Zona_1'] > limiar_70]
total_amostra = len(df)
qtd_df1 = len(df1)
pct_df1 = (qtd_df1 / total_amostra) * 100

print('=== 5. Contagem e Percentual (DataFrame 1) ===')
print(f'Total de registros na amostra: {total_amostra}')
print(f'Registros acima do limiar (df1): {qtd_df1}')
print(f'Percentual da amostra total: {pct_df1:.2f}%\n')
```

```python
# 6. Temperatura Média e DataFrame 2
temp_media = df['Temperature'].mean()
df2 = df1[df1['Temperature'] > temp_media]
qtd_df2 = len(df2)
pct_df2 = (qtd_df2 / total_amostra) * 100

print('=== 6. Temperatura Média e DataFrame 2 ===')
print(f'Temperatura Média Geral: {temp_media:.2f} °C')
print(
    f'Registros filtrados em df2 (Consumo > {limiar_70:,.2f} kW E Temp > {temp_media:.2f} °C): {qtd_df2}'
)
print(f'Percentual da amostra total: {pct_df2:.2f}%\n')
```

```python
# 7. Comparação e Explicação
reducao = qtd_df1 - qtd_df2
pct_reducao = (reducao / qtd_df1) * 100
pct_coincide = (qtd_df2 / qtd_df1) * 100

print('=== 7. Comparação e Explicação ===')
print(
    f'Redução: A inclusão do filtro de temperatura reduziu o conjunto de {qtd_df1} para {qtd_df2} registros.'
)
print(
    f'Diferença: Queda de {reducao} registros (~{pct_reducao:.1f}% dos registros de pico).\n'
)
print('Análise:')
print(
    f' - {pct_coincide:.1f}% dos picos da Zona 1 coincidem com temperaturas acima da média ({temp_media:.2f} °C),'
)
print(
    '   o que aponta padrão típico do uso de equipamentos de refrigeração e ar-condicionado.'
)
print(
    f' - {pct_reducao:.1f}% dos picos ocorrem sob temperaturas amenas/baixas,'
)
print(
    '   sendo induzidos por outros fatores de demanda industrial, comercial ou operacional.'
)
```

## Dataset 5 — Wind & Solar Energy Production (Kaggle)


#####Fonte: https://www.kaggle.com/datasets/ahmeduzaki/wind-and-solar-energy-production-dataset

```python
file_path = "Energy Production Dataset.csv"
df = pd.read_csv(file_path)

print("=== ETAPA A.1: Atributos disponíveis ===")
print("Colunas:", df.columns.tolist())
print("\nPrimeiras 5 linhas:")
print(df.head())

colunas_temporais = [
    "Date",
    "Start_Hour",
    "End_Hour",
    "Day_of_Year",
    "Day_Name",
    "Month_Name",
    "Season",
]
colunas_selecionadas = colunas_temporais + ["Source", "Production"]
df_selecionado = df[colunas_selecionadas].copy()

print("\n=== ETAPA A.2: Atributos Mantidos ===")
print(df_selecionado.head())

solar_min = df_selecionado[df_selecionado["Source"] == "Solar"][
    "Production"
].min()
solar_max = df_selecionado[df_selecionado["Source"] == "Solar"][
    "Production"
].max()

eolica_min = df_selecionado[df_selecionado["Source"] == "Wind"][
    "Production"
].min()
eolica_max = df_selecionado[df_selecionado["Source"] == "Wind"][
    "Production"
].max()

print("\n=== ETAPA A.3: Valores Mínimos e Máximos ===")
print(f"Geração Solar  -> Mínimo: {solar_min} | Máximo: {solar_max}")
print(f"Geração Eólica -> Mínimo: {eolica_min} | Máximo: {eolica_max}")

print("\n=== ETAPA A.4: Valores Ausentes ===")
print(df_selecionado.isnull().sum())

df_amostra = df_selecionado.sample(frac=0.20, random_state=42)

df_amostra.to_csv("amostra_20_percent.csv", index=False)
print(
    f"\n=== ETAPA A.5 & A.6: Amostra gerada com {len(df_amostra)} registros e salva em 'amostra_20_percent.csv' ==="
)

df_solar = (
    df_selecionado[df_selecionado["Source"] == "Solar"]
    .copy()
    .rename(columns={"Production": "Geracao_Solar"})
)
df_eolica = (
    df_selecionado[df_selecionado["Source"] == "Wind"]
    .copy()
    .rename(columns={"Production": "Geracao_Eolica"})
)

max_solar = df_solar["Geracao_Solar"].max()
max_eolica = df_eolica["Geracao_Eolica"].max()

print("\n=== ETAPA B.2: Valores Máximos Registrados ===")
print(f"Máximo Solar : {max_solar} MW")
print(f"Máximo Eólico: {max_eolica} MW")

limite_solar_70 = max_solar * 0.70
limite_eolico_70 = max_eolica * 0.70

print("\n=== ETAPA B.3: Limites de 70% do Máximo ===")
print(f"Limite 70% Solar : {limite_solar_70:.2f} MW")
print(f"Limite 70% Eólico: {limite_eolico_70:.2f} MW")

df_alta_solar = df_solar[df_solar["Geracao_Solar"] >= limite_solar_70]
df_alta_eolica = df_eolica[df_eolica["Geracao_Eolica"] >= limite_eolico_70]

total_registros_solar = len(df_solar)
total_registros_eolica = len(df_eolica)

qtd_alta_solar = len(df_alta_solar)
qtd_alta_eolica = len(df_alta_eolica)

pct_alta_solar = (qtd_alta_solar / total_registros_solar) * 100
pct_alta_eolica = (qtd_alta_eolica / total_registros_eolica) * 100

print("\n=== ETAPA B.5: Frequência e Percentuais de Alta Geração ===")
print(
    f"Alta Geração Solar  (>= 70%): {qtd_alta_solar} registros de {total_registros_solar} ({pct_alta_solar:.2f}%)"
)
print(
    f"Alta Geração Eólica (>= 70%): {qtd_alta_eolica} registros de {total_registros_eolica} ({pct_alta_eolica:.2f}%)"
)

print("\n=== ETAPA B.6: Comparação de Frequência ===")
if pct_alta_solar > pct_alta_eolica:
    print(
        f"A fonte SOLAR apresentou maior frequência proporcional acima de 70% do seu topo ({pct_alta_solar:.2f}% vs {pct_alta_eolica:.2f}%)."
    )
elif pct_alta_eolica > pct_alta_solar:
    print(
        f"A fonte EÓLICA apresentou maior frequência proporcional acima de 70% do seu topo ({pct_alta_eolica:.2f}% vs {pct_alta_solar:.2f}%)."
    )
else:
    print("Ambas as fontes possuem exatamente o mesmo percentual relativo.")
```

**Situação**

```python
import io
import zipfile
import pandas as pd
from google.colab import files

print("=== PIPELINE COMPLETO (ETAPAS A e B) ===")

# ---------------------------------------------------------
# ETAPA A: UPLOAD, SELEÇÃO, TRATAMENTO E AMOSTRAGEM
# ---------------------------------------------------------
print("\n1. Faça o upload do arquivo bruto (.csv, .txt ou .zip):")
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

# Tratamento para extrair arquivo caso seja enviado em formato .ZIP
if file_name.endswith(".zip"):
  with zipfile.ZipFile(io.BytesIO(uploaded[file_name]), "r") as z:
    # Pega o primeiro arquivo dentro do zip que seja csv ou txt
    internal_files = [
        f for f in z.namelist() if f.endswith(".csv") or f.endswith(".txt")
    ]
    extracted_file = z.open(internal_files[0])
    # Tenta ler com UTF-8 e fallback para ISO-8859-1 se houver caracteres especiais
    try:
      df_raw = pd.read_csv(extracted_file, sep=None, engine="python")
    except UnicodeDecodeError:
      extracted_file.seek(0)
      df_raw = pd.read_csv(
          extracted_file, sep=None, engine="python", encoding="iso-8859-1"
      )
else:
  # Carregamento direto caso não seja zip
  try:
    df_raw = pd.read_csv(
        io.BytesIO(uploaded[file_name]),
        sep=";",
        na_values="?",
        low_memory=False,
    )
  except Exception:
    df_raw = pd.read_csv(
        io.BytesIO(uploaded[file_name]), sep=",", low_memory=False
    )

# Mapeamento dinâmico de colunas (Atende tanto Household Power quanto Appliances Energy)
colunas_household = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3",
]
colunas_appliances = [
    "Appliances",
    "lights",
    "T1",
    "RH_1",
    "T2",
    "RH_2",
    "Press_mm_hg",
]

if all(col in df_raw.columns for col in colunas_household):
  colunas_desejadas = colunas_household
elif all(col in df_raw.columns for col in colunas_appliances):
  colunas_desejadas = colunas_appliances
else:
  # Caso as colunas tenham nomes ligeiramente diferentes, seleciona as 7 primeiras numéricas
  colunas_desejadas = df_raw.select_dtypes(include=["number"]).columns[:7].tolist()

df = df_raw[colunas_desejadas].copy()

# Imputação de nulos usando a mediana de cada coluna
for col in df.columns:
  df[col] = pd.to_numeric(df[col], errors="coerce")
  df[col] = df[col].fillna(df[col].median())

# Amostra aleatória de 10%
df_amostra = df.sample(frac=0.10, random_state=42).copy()

# Salvando a amostra tratada em CSV
df_amostra.to_csv("amostra_household_10pct.csv", index=False)

# ---------------------------------------------------------
# ETAPA B: RENOMEAÇÃO, CÁLCULOS E FILTRAGENS
# ---------------------------------------------------------
# Normalizando os nomes para as 7 colunas do exercício
df_amostra.columns = [
    "pot_ativa",
    "pot_reativa",
    "tensao",
    "corrente",
    "sub_1",
    "sub_2",
    "sub_3",
]

# Determinação da Potência Máxima e do Limite de 75%
max_pot_ativa = df_amostra["pot_ativa"].max()
limite_75_pot = max_pot_ativa * 0.75

# DataFrame 1: Potência Ativa > 75% do máximo
df_demanda_alta = df_amostra[df_amostra["pot_ativa"] > limite_75_pot]
qtd_demanda_alta = len(df_demanda_alta)
pct_demanda_alta = (qtd_demanda_alta / len(df_amostra)) * 100

# Cálculo da Corrente Média da amostra
corrente_media = df_amostra["corrente"].mean()

# DataFrame 2: Potência Ativa > 75% E Corrente > Média
df_critico = df_amostra[
    (df_amostra["pot_ativa"] > limite_75_pot)
    & (df_amostra["corrente"] > corrente_media)
]
qtd_critico = len(df_critico)
pct_critico = (qtd_critico / len(df_amostra)) * 100

# ---------------------------------------------------------
# RELATÓRIO FINAL DE SAÍDA
# ---------------------------------------------------------
print("\n" + "=" * 45)
print("             RELATÓRIO DE RESULTADOS         ")
print("=" * 45)
print(f"Total de registros originais:    {len(df_raw)}")
print(f"Total na amostra de 10%:         {len(df_amostra)}")
print(f"Potência Ativa Máxima (Amostra): {max_pot_ativa:.3f}")
print(f"Limite de Corte (75% do Máx):    {limite_75_pot:.3f}")
print(f"Corrente Média da Amostra:       {corrente_media:.2f}\n")

print(
    f"1. Demanda Alta (Pot. Ativa > 75%):"
    f"\n   -> {qtd_demanda_alta} registros ({pct_demanda_alta:.2f}% da amostra)\n"
)

print(
    f"2. Demanda Crítica (Pot. Ativa > 75% E Corrente > Média):"
    f"\n   -> {qtd_critico} registros ({pct_critico:.2f}% da amostra)"
)
print("=" * 45)
```

A relação básica de potência em sistemas elétricos é aproximada por **Potência =~ Tensão x Corrente**.

Eliminação de Oscilações de Tensão: Momentos em que a potência atinge valores elevados podem ser causados tanto por alta corrente quanto por variações/picos na tensão fornecida pela rede. Exigir que a corrente também seja superior à média garante que o evento corresponda a uma demanda física real de carga (vários aparelhos ou equipamentos de alta potência ligados simultaneamente).
Confirmação de Carga Crítica: Na prática, esse filtro combinado reduz o ruído de medições isoladas e confirma episódios em que a fiação e os disjuntores da residência estão sendo submetidos a uma operação de alta carga contínua.
